```python
xtrain=[]
ytrain=[]
```

```python
datadir=('/content/drive/MyDrive/Colab Notebooks/Mask data (1)/Mask_train')
```

```python
categories = sorted(os.listdir('/content/drive/MyDrive/Colab Notebooks/Mask data (1)/Mask_train'))
categories
```

```python
for i in categories:
    path = os.path.join(datadir, i)
    for img in os.listdir(path):
        imgpath = os.path.join(path, img)
        img1 = imread(imgpath)
        img1_res = resize(img1, (150, 150, 3))
        xtrain.append(img1_res)
        ytrain.append(categories.index(i))
```

testing


```python
xtest=[]
ytest=[]
```

```python
datadir1 = "/content/drive/MyDrive/Colab Notebooks/Mask data (1)/Mask_test"
```

```python
categories1 = sorted(os.listdir("/content/drive/MyDrive/Colab Notebooks/Mask data (1)/Mask_test"))
categories1
```

```python
for j in categories1:
  path1 = os.path.join(datadir1, j)

  for imgs in os.listdir(path1):
    img_path1 = os.path.join(path1, imgs)

    img2 = imread(img_path1)

    img_res1 = resize(img2, (150, 150, 3))

    xtest.append(img_res1)

    ytest.append(categories1.index(j))
```

```python
xtrain = np.array(xtrain)
ytrain = np.array(ytrain)
xtest = np.array(xtest)
ytest = np.array(ytest)
```

```python
print("Training data shape:", xtrain.shape)
print("Training labels shape:", ytrain.shape)

print("Testing data shape:", xtest.shape)
print("Testing labels shape:", ytest.shape)
```

```python
model = Sequential()
```

layers

```python
model.add(Conv2D(32, (3, 3), activation = "relu", input_shape = (150, 150 ,3)))

model.add(MaxPool2D(2,2))

model.add(Flatten())

model.add(Dense(60, activation = "relu"))
model.add(Dense(30, activation = "relu"))

model.add(Dense(1, activation = "sigmoid"))
```

```python
# compile
model.compile(optimizer = "adam",
              loss = "binary_crossentropy",
              metrics = ["accuracy"])
```

```python
# training
model.fit(xtrain, ytrain, epochs = 30, batch_size = 50)
```

```python
# prediction
```

```python
image_paths = "/content/drive/MyDrive/Colab Notebooks/Mask data (1)/Mask_train/with_mask/1-with-mask.jpg"

img = imread(image_paths)

img_resized = resize(img, (150, 150, 3))

prediction = model.predict(np.array([img_resized]))

if prediction[0][0] >= 0.5:
    print("Without Mask")
else:
    print("With Mask")
```
