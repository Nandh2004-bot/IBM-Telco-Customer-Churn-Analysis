```python
# iqr monthly charge
Q1 = df['MonthlyCharges'].quantile(0.25)
Q3 = df['MonthlyCharges'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', IQR)
print('Lower limit:', lower)
print('Upper limit:', upper)
```

```python
outliers = df[
    (df['MonthlyCharges'] < lower) |
    (df['MonthlyCharges'] > upper)
]

print('Number of outliers:', len(outliers))
```

```python
ax = sns.boxplot(data=df, y='MonthlyCharges')

ax.set_title('Outlier Detection in Monthly Charges')
ax.set_ylabel('Monthly Charges')

plt.show()
```

```python
# Outlier detection in Tenure
Q1 = df['tenure'].quantile(0.25)
Q3 = df['tenure'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['tenure'] < lower) | (df['tenure'] > upper)]

print('Number of outliers:', len(outliers))
```

```python
# Outlier detection in TotalCharges

Q1 = df['TotalCharges'].quantile(0.25)
Q3 = df['TotalCharges'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['TotalCharges'] < lower) | (df['TotalCharges'] > upper)]

print('Number of outliers:', len(outliers))
```

```python

# How strongly are the numerical variables related to each other?

corr = df[['tenure', 'MonthlyCharges', 'TotalCharges']].corr()

plt.figure(figsize=(8, 5))

ax = sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')

ax.set_title('Correlation Heatmap')

plt.show()
```
